```python
sns.boxplot(data = df, x = 'Category', y = 'Profit')
plt.title('Category VS Profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

```python
sns.boxplot(data = df, x = 'Region', y = 'Sales')
plt.title('Region VS Sales')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
sns.boxplot(data = df, x = 'Region', y = 'Profit')
plt.title('Region VS Profit')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
```

### Conclusion

The bivariate analysis reveals noticeable relationships between sales, profit, discount, categories, and regions. Sales and profit show a generally positive relationship, although several high-value transactions behave differently from the overall pattern. Higher discount levels are associated with lower profit in many observations. The category-wise analysis shows considerable variation in sales and profit, while regional differences are also visible. These findings indicate that discount, category, and region may have an important impact on overall profitability and require further business-level analysis.

## Business-Level Analysis

```python
df.head()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()
category_sales
```

```python
plt.figure(figsize=(8, 5))

category_sales.plot(kind='bar', color='steelblue')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum()
category_profit
```

```python
plt.figure(figsize=(8, 5))

category_profit.plot(kind='bar', color = 'blue')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)

plt.show()
```

### Conclusion

The category-wise analysis shows that Technology generated the highest total sales, followed by Furniture, while Office Supplies recorded the lowest total sales. The profit analysis also highlights differences in profitability across the three categories, indicating that higher sales do not necessarily imply proportionally higher profit. Therefore, both sales and profit should be considered when evaluating category performance.
