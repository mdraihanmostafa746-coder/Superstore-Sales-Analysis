## Univariate Analysis

```python
df.head(5)
```

```python
sns.histplot(
    df['Sales'], 
    color='steelblue'
)


plt.show()
```

```python
df.boxplot(
    'Sales', 
    color='steelblue'
)


plt.show()
```

```python
sns.histplot(
    df['Quantity'], 
    color = 'darkorange'
)


plt.show()
```

```python
sns.countplot(
    data=df, 
    x='Quantity', 
    color = "darkorange" 
)


plt.show()
```

```python
sns.histplot(
    df['Discount'], 
    color = 'mediumpurple'
)


plt.show()
```

```python
sns.histplot(
    df['Profit'], 
    color = 'seagreen'
)


plt.show()
```

```python
df.boxplot(
    'Profit', 
    color = 'seagreen'
)


plt.show()
```

```python
df.head(3)
```

```python
sns.countplot(
    data=df, 
    x='Category', 
    palette='Set2' 
)


plt.show()
```

```python
sns.countplot(
    data=df, 
    x='Segment',
    palette='Set1' 
)


plt.show()
```

```python
sns.countplot(
    data=df,
    x='Ship Mode', 
    palette='Set1'
)


plt.show()
```

```python
sns.countplot(
    data=df, 
    x='Sub-Category', 
    palette='Set1' 
)


plt.xticks(rotation=90, fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    y='Sub-Category',
    data=df,
    order=df['Sub-Category'].value_counts().index,
    color='steelblue'
)

plt.title('Number of Orders by Sub-Category')
plt.xlabel('Number of Orders')
plt.ylabel('Sub-Category')

plt.show()
```

### Conclusion

The univariate analysis shows that the Superstore dataset contains a highly right-skewed Sales distribution with several high-value transactions. Profit also varies considerably across orders, while Quantity and Discount have more limited ranges. The categorical analysis shows that order distribution is not uniform across categories, sub-categories, segments, regions, and shipping modes. These observations provide a basic understanding of the dataset and help identify patterns to investigate further through bivariate analysis.
